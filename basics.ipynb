```python
import numpy as np
```

```python
s = 5 # Scalar
v = np.array([1,2,3]) # Vector(1D array)
m = np.array([[1,2],[3,4],[5,6]]) # Matrix (2D array)

print(f"Scalar : {s}")
print(f"Vector : {v}")
print(f"Matrix : \n{m}")
```

```python
# Vectors
v1 = np.array([1,2])
v2 = np.array([3,4])
# Scalar multiplication
scalar_v = s*v1
print(f"Scalar - Vector Multiplication: {scalar_v}")
# Vector addition
v_sum = v1 + v2
print(f"Vector addition result: {v_sum}")
# Dot product
dot_prod = np.dot(v1, v2)
print(f"Dot product of {v1} and {v2} : {dot_prod}")

# Matrices
A = np.array([[4,3],[3,2]])
B = np.array([[5,6],[7,8]])
# Vector Multiplication
vect_matrix_mult = A @ v1
print(f"Vector - Matrix multiplication: {vect_matrix_mult}")
# Matrix Multiplication
matrix_mult = A @ B
print(f"Matrix - Matrix multiplication: {matrix_mult}")
# Matrix Transpose
A_transpose = A.T
print(f"Transpose of matrix A is : {A_transpose}")
# Matrix Inverse
A_inverse = np.linalg.inv(A)
print(f"Matrix Inverse of A is : {A_inverse}")
# Matrix - Matrix inverse multiplication result to identity matrix
print(f"Matrix and Matrix-Inverse multiplication: {A @ A_inverse}")
```

```python
# Finding roots of x^2 - 5x + 6 = 0 using NumPy
coefficients = [1, -5, 6]
roots = np.roots(coefficients) 
print(f"roots of quadratic equation X^2 - 5x + 6 = 0 is:{roots}")
```

```python
# Calculating Eigenvalues (w) and Eigenvectors (v)
A = np.array([[4, 2], [1, 3]])
eigenvalues, eigenvectors = np.linalg.eig(A)
print(f"Eigen Values: {eigenvalues}")
print(f"Eigen Vector: {eigenvectors}")
```
