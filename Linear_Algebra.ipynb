```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("/content/hdfc_loan_dataset_full_enriched - hdfc_loan_dataset_full_enriched.csv")
```

```python
data
```

```python
data.info()
```

```python
# Vector Addition to find applicant household income per month
applicant_household_income = data['Applicant_Income'] + data['Coapplicant_Income']
print(applicant_household_income)
```

```python
# Monthly saving vector subtraction
monthly_saving = applicant_household_income - data['Monthly_Expense']
print(monthly_saving)
```

```python
# Predicting fixed expense : Matrix- vector multiplication
w = np.array([100,1])
matrix = data[['Dependents','Existing_EMIs']].to_numpy()
predicted_fixed_expense = np.dot(matrix,w)
print(predicted_fixed_expense)
```

```python
# scalar - vector multiplication
# if by any chance Applicant income rise by 10 %
predicted_income = data['Applicant_Income'] * 1.10
print("Predicted Income:\n",predicted_income)
```

```python
# vectors - dot product
# Monthly intrest bank earns
loans = data[data['Loan_Status'] == 'Approved']['Loan_Amount'].to_numpy()
intrest_rate = np.ones(len(loans))*0.12 # 12 % intrest
monthly_intrest = np.dot(loans,intrest_rate)
print(monthly_intrest)
```
